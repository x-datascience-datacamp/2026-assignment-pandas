```python
df1 = pd.read_csv("data/regions.csv")
df2 = pd.read_csv("data/departments.csv")
```

```python
import pandas as pd

# 1. Prepare df2 (Regions): Keep only 'id', rename the rest immediately
regions = df1[["id", "code", "name"]].rename(
    columns={"code": "code_reg", "name": "name_reg"}
)

# 2. Prepare df1 (Departments): Keep only 'id', rename the rest immediately
departments = df2[["id", "code", "name"]].rename(
    columns={"code": "code_dep", "name": "name_dep"}
)

# 3. Merge them together on 'id' and drop 'id' right away
final_df = pd.merge(regions, departments, on="id").drop(columns="id")

final_df
```
